### Ratings Full-file Check

One scan for duplicates, missing values, types, categories, and helpfulness. Earlier Ratings checks use the 1,000-row sample.


```python
import sqlite3
import tempfile
from pathlib import Path

ratings_path = "~/Downloads/ratings-00006.tsv"

total_rows = 0
missing_counts = pd.Series(0, index=ratings.columns, dtype="int64")
categorical_values = {
    "helpfulnessLevel": set(),
    "ratingSourceBucketed": set()
}
rating_types = {col: set() for col in ratings.columns}
duplicate_pairs = 0
valid_ratings = 0
helpful_ratings = 0

with tempfile.TemporaryDirectory() as temp_dir:
    with sqlite3.connect(str(Path(temp_dir) / "ratings.sqlite")) as conn:
        conn.execute("""
            CREATE TABLE pairs (
                noteId TEXT,
                raterParticipantId TEXT,
                PRIMARY KEY (noteId, raterParticipantId)
            ) WITHOUT ROWID
        """)

        for chunk in pd.read_csv(
            ratings_path,
            sep="\t",
            chunksize=100000,
            dtype={
                "noteId": "string",
                "raterParticipantId": "string",
                "suggestion": "string"
            }
        ):
            total_rows += len(chunk)
            missing_counts += chunk.isna().sum()

            for col in chunk.columns:
                rating_types[col].add(str(chunk[col].dtype))

            for col in categorical_values:
                categorical_values[col].update(chunk[col].dropna().unique())

            pairs = chunk[["noteId", "raterParticipantId"]].dropna()
            before = conn.total_changes
            conn.executemany(
                "INSERT OR IGNORE INTO pairs VALUES (?, ?)",
                pairs.itertuples(index=False, name=None)
            )
            duplicate_pairs += len(pairs) - (conn.total_changes - before)
            conn.commit()

            valid = chunk["helpfulnessLevel"].isin(
                ["HELPFUL", "SOMEWHAT_HELPFUL", "NOT_HELPFUL"]
            )
            valid_ratings += int(valid.sum())
            helpful_ratings += int(chunk["helpfulnessLevel"].eq("HELPFUL").sum())

print("Total rows:", total_rows)
print("Duplicate noteId + raterParticipantId:", duplicate_pairs)
print("Missing key fields:")
print(missing_counts[["noteId", "raterParticipantId"]])
print("Missing before/after (values preserved):")
print(pd.DataFrame({"before": missing_counts, "after": missing_counts}))
print("Types:", rating_types)
print("Categorical values:", categorical_values)
print("Helpfulness denominator:", valid_ratings)
print("Excluded missing/invalid helpfulness:", total_rows - valid_ratings)
print("Helpful ratings:", helpful_ratings)
print("Observed helpfulness:",
      helpful_ratings / valid_ratings if valid_ratings else float("nan"))

if duplicate_pairs == 0 and missing_counts[
    ["noteId", "raterParticipantId"]
].sum() == 0:
    print("Full-file exact duplicate rows: 0 (unique complete keys)")
else:
    print("Exact-row duplicate verification still required.")
```

## 11. Check Calculations

Helpfulness uses all valid Ratings values from section 7.

Coverage is checked separately for each feed through October 2, 2026, 23:59:59.999 UTC. Each post is counted once. A matching note must be created after eligibility and by the cutoff. Eligibility values of -1 are excluded. This cutoff is used for verification.

Response time excludes missing timestamps and negative durations. The main measure is the median. These are overall calculations; classification is not used to group records.


### Response Time: Mean


```python
# Response time for Notes linked to Status
response_time = (
    notes_status["timestampMillisOfFirstNonNMRStatus"]
    - notes_status["createdAtMillis_x"]
) / 3600000

valid = response_time.notna() & response_time.ge(0)

print("Notes examined:", len(notes_status))
print("Missing response time:", response_time.isna().sum())
print("Negative response time:", response_time.lt(0).sum())
print("Response-time denominator:", valid.sum())
print("Mean response time hours:", response_time[valid].mean())
```

### Coverage and Response Time: Median


```python
# Verification cutoff: end of October 2, 2026 (UTC)
cutoff = pd.Timestamp("2026-10-03", tz="UTC").value // 10**6 - 1

for col in [
    "noteRequestFeedEligibleAtMillis",
    "apiSmallFeedEligibleAtMillis",
    "apiLargeFeedEligibleAtMillis",
    "apiXlFeedEligibleAtMillis"
]:
    eligible = requests[
        requests["tweetId"].gt(0)
        & requests[col].ge(0)
        & requests[col].le(cutoff)
    ]["tweetId"].nunique()

    covered = requests_notes[
        requests_notes["tweetId"].gt(0)
        & requests_notes[col].ge(0)
        & requests_notes[col].le(cutoff)
        & requests_notes["createdAtMillis"].gt(requests_notes[col])
        & requests_notes["createdAtMillis"].le(cutoff)
    ]["tweetId"].nunique()

    print(col)
    print("Eligible posts:", eligible)
    print("Covered posts:", covered)
    print("Coverage gap:", eligible - covered)
    print("Coverage:", covered / eligible if eligible else float("nan"))

response_time = (
    notes_status["timestampMillisOfFirstNonNMRStatus"]
    - notes_status["createdAtMillis_x"]
) / 3600000

valid = (
    response_time.notna()
    & response_time.ge(0)
    & notes_status["createdAtMillis_x"].le(cutoff)
    & notes_status["timestampMillisOfFirstNonNMRStatus"].le(cutoff)
)

print("Response-time denominator:", valid.sum())
print("Median response time hours:", response_time[valid].median())
print("Missing classification among included notes:",
      notes_status.loc[valid, "classification"].isna().sum())
```
